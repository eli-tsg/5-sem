```python
import numpy as np
from scipy import stats

# среднее выборочное
def mean(data):
    return np.mean(data)

# СКО
def std(data):
    return np.std(data)

# хи-квадрат
def chi2(a, df):
    return stats.chi2.ppf(a, df)

# t-коэффициент
def t_coeff(a, df):
    return stats.t.ppf(1 - a / 2, df)

def conf_interval(x_mean, x_std, x_y, N):
    left_border = x_mean - x_y * (x_std / (N ** 0.5))
    right_border = x_mean + x_y * (x_std / (N ** 0.5))
    return (left_border, right_border)

def conf_interval_2(x_mean, x_std, t_coeff, N):
    left_border = x_mean - t_coeff * (x_std / (N ** 0.5))
    right_border = x_mean + t_coeff * (x_std / (N ** 0.5))
    return (left_border, right_border)

def conf_interval_std(x_std, N, chi2_add, chi2_sub):
    left_border = (x_std * (N - 1) ** 0.5) / ((chi2_add) ** 0.5)
    right_border = (x_std * (N - 1) ** 0.5) / ((chi2_sub) ** 0.5)
    return (left_border, right_border)
```

```python
def find_intervals(x_mean, x_std, x_y, x_t_coeff, x_chi2_add, x_chi2_sub, N):
    ci_ev_1 = conf_interval(x_mean, x_std, x_y, N)
    ci_ev_2 = conf_interval_2(x_mean, x_std, x_t_coeff, N)
    ci_std = conf_interval_std(x_std, N, x_chi2_add, x_chi2_sub)
    answers = {'Доверительный интервал для точечной оценки выборочного среднего':ci_ev_1, 
            'Итоговый доверительный интервал для точечной оценки выборочного среднего':ci_ev_2,
            'Доверительный интервал для точечной оценки СКО':ci_std}
    return answers
```

```python
directory = 'вариант29'
names = ['1.txt','2.txt', '3.txt', '4.txt']

for name in names:
    filename = directory + '/' + name
    with open(filename, 'r') as file:
        data = [float(x) for x in file]
    print("=" * 80)
    print(f'ФАЙЛ: {name}')
    N = len(data)
    df = N - 1
    y = 0.95
    a = 1 - y
    x_y = 1.96
    x_mean = mean(data)
    x_std = std(data)
    print(f'Выборочное среднее = {x_mean:.3f}')
    print(f'Выборочное СКО = {x_std:.3f}')
    x_t_coeff = t_coeff(a, df)
    a1, a2 = round((1 + y) / 2, 3), round((1 - y) / 2, 3)
    x_chi2_add = chi2(a1, df)
    x_chi2_sub = chi2(a2, df)
    print(f'Значение t-распределения = {x_t_coeff:.3f}')
    print(f'Значение хи2 для ({a1}, {df}) = {x_chi2_add:.3f}')
    print(f'Значение хи2 для ({a2}, {df}) = {x_chi2_sub:.3f}')
    print()
    intervals = find_intervals(x_mean, x_std, x_y, x_t_coeff, x_chi2_add, x_chi2_sub, N)
    for key, value in intervals.items():
        print(key)
        l_border, r_border = value
        print(f'[{l_border:.3f}; {r_border:.3f}]')
        print()
```
