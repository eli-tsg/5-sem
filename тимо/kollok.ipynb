```python
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
import numpy as np
x = [1, 2, 3, 4, 5]
y = [1, 4, 9, 16, 25]
def find_coeffs(x, y):
    poly = PolynomialFeatures(degree=2)
    x_2d = np.array(x).reshape(-1, 1)
    x_poly = poly.fit_transform(x_2d)
    linreg = LinearRegression()
    linreg.fit(x_poly, y)
    return linreg.coef_

print(find_coeffs(x, y))
```

```python
def sliding_window(x, w, step):
    slides = []
    for i in range(0, len(x) - w + 1, step):
        slides.append(x[i:i + w])
    return slides
window = 3
step_s = 1
x1 = np.array([8, 1, 4, 5, -2, 5, 9, 0])
print(sliding_window(x1, w=window, step=step_s))
```

```python
from numpy import sqrt
v1 = [1, 0, 1]
v2 = [0, 1, 1]
def cosine(v1, v2):
    a = sum([v1[i] * v2[i] for i in range(len(v1))])
    b = sqrt(sum(i ** 2 for i in v1))
    c = sqrt(sum(i ** 2 for i in v2))
    return a / (b * c)
print(cosine(v1, v2))
```

```python
import numpy as np
data = [1, 2, 3, 4, 100]
def calculate_mad(data):
    data = np.array(data)
    median = np.median(data)
    abs_deviations = np.abs(data - median)
    mad = np.median(abs_deviations)
    return mad
print(calculate_mad(data))
```
